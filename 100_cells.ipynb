## Раздел 1. Имя и значение

`=` — это не «равно». Это приказ: **справа посчитай, слева положи**.

```
левая часть  =  правая часть
   куда           что
```

````python
# 1 — что напечатается?
x = 5
print(x)
````

````python
# 2 — а теперь?
x = 5
x = 10
print(x)
````

Второе присваивание не создало новое имя, а **перезаписало** старое. Имя одно, значение новое.

````python
# 3 — сначала считается правая часть
x = 2 + 3
print(x)
````

````python
# 4 — то же самое, но в два шага
2 + 3
````

В Jupyter последняя строка ячейки показывает своё значение сама, без `print`. Это удобно для опытов.

````python
# 5 — что окажется в y?
x = 5
y = x
print(y)
````

````python
# 6 — предскажи ОБА числа
x = 5
y = x
x = 10
print(x)
print(y)
````

`y = x` не привязало `y` к `x` навсегда. Питон посмотрел, что **сейчас** лежит в `x` (пятёрка),
и положил пятёрку в `y`. Дальше они живут отдельно.

````python
# 7 — имя x и строка "x" — разные вещи
x = 5
print(x)
print("x")
````

````python
# 8 — всё это одно и то же действие
a = ""
A = ""
aaa = ""
AAAAAAA = ""
print(a, A, aaa, AAAAAAA)
````

````python
# 9 — но a и A — РАЗНЫЕ имена
a = 1
A = 2
print(a)
print(A)
````

````python
# 10 — одно равно — «сделай так»
x = 5
print(x)
````

````python
# 11 — два равно — «а так ли это?»
x = 5
print(x == 5)
print(x == 7)
````

````python
# 12 — += это «прибавь к тому, что уже есть»
our_number = 42
our_number = our_number + 1
print(our_number)
our_number += 2
print(our_number)
````

**Где мы это видели:** `week1/learn_syntax.py` — переменные, `our_number += 2`, секунды в году.

## Раздел 7. Косвенность: ключ в переменной

Здесь начинается место, где «всё было понятно» и вдруг перестало.
Синтаксис почти не меняется, а читать становится трудно.

Лекарство одно: **не читать строку целиком. Спрашивать, чему равен каждый кусочек.**

````python
# 62 — ключ можно положить в переменную
a = {"red": 1, "blue": 0, "orange": "Оранжевый"}
c = "red"
print(c)
print(a[c])
````

Разворачиваем по шагам:

```
c        ->  "red"
a[c]     ->  a["red"]
         ->  1
```

Питон сначала вычисляет то, что **внутри** скобок, и только потом лезет в словарь.

````python
# 63 — это одно и то же
print(a["red"])
print(a[c])
print(c == "red")
````

````python
# 64 — поменяли c — и a[c] стало другим
c = "orange"
print(a[c])
````

````python
# 65 — так работает морзянка: letter — переменная
morse_dict = {"A": ".-", "L": ".-..", "F": "..-."}
for letter in "ALFA":
    print(letter, morse_dict[letter])
````

В цикле `letter` каждый раз новое, поэтому `morse_dict[letter]` каждый раз достаёт другое значение. Это та же косвенность.

````python
# 66 — значением может быть список из других имён
b = {"name": "Bob"}
c = 10
d = "hello"

a = {"id1": [b, c, d]}
print(a)
````

В списке лежат `b`, `c`, `d` — это **имена**, а не буквы. Питон посмотрел, что в них лежит,
и положил в список именно значения: словарь, число, строку.

````python
# 67 — сравни: имя b и строка "b"
print(b)
print("b")
````

````python
# 68 — разбираем a["id1"][0]["name"] по слоям
print(a["id1"])
print(a["id1"][0])
print(a["id1"][0]["name"])
````

````python
# 69 — словарь внутри словаря
a = {"x": {"red": 1, "blue": 0}}
print(a["x"])
print(a["x"]["red"])
````

````python
# 70 — список словарей внутри словаря
a = {"x": [{"red": 1}, {"blue": 0}]}
print(a["x"])
print(a["x"][0])
print(a["x"][0]["red"])
````

````python
# 71 — то же самое, но ключ в переменной
c = "x"
print(a[c])
print(a[c][0])
print(a[c][0]["red"])
````

````python
# 72 — и наконец — присваивание внутрь
a[c][0]["red"] = 2
print(a)
````

Слева от `=` тоже всё вычисляется по слоям:

```
a[c]            ->  [{"red": 1}, {"blue": 0}]
a[c][0]         ->  {"red": 1}
a[c][0]["red"] = 2   ->  в этом словаре по ключу "red" кладём 2
```

Ничего нового здесь не появилось. Это цепочка из операций, которые мы уже знаем.

`if` не понимает «условие». `if` понимает ровно одну вещь: **правда или ложь**.
Всё, что стоит после `if`, сначала превращается в `True` или `False`.

````python
# 78 — сравнение возвращает True или False
print(5 == 5)
print(5 != 10)
print(10 > 20)
print(type(5 == 5))
````

````python
# 79 — сравнения можно ставить цепочкой
print(6 < 10 < 15)
print(6 <= 6 <= 20)
````

````python
# 81 — результат сравнения можно сохранить в переменную
is_big = 10 > 20
print(is_big)

if is_big:
    print("большое")
else:
    print("маленькое")
````

````python
# 82 — Питон считает «правдой» не только True
if 2:
    print("двойка — это правда")

if 0:
    print("ноль — это правда")

print("конец")
````

## Раздел 9. None

`None` нужен там, где значения **нет** — и это не то же самое, что ноль или пустая строка.

````python
# 91 — 0 и None — разные вещи
temperature = 0
print(temperature)

temperature = None
print(temperature)
print(type(None))
````

`temperature = 0` значит «измерили, там ноль градусов».
`temperature = None` значит «не измерили, не знаем».

Заменить `None` нулём нельзя: ноль — это законное значение со своим смыслом.

````python
# 92 — None не равен 0, "" и False
print(None == 0)
print(None == "")
print(None == False)
````

## Раздел 10. Цикл

Цикл — это не новая магия. Это сжатая запись повторяющихся строк.

> Цикл берёт значения по одному, кладёт очередное в переменную и выполняет один и тот же блок.

````python
# 101 — имя переменной цикла придумываем мы — Питону всё равно
for cube in range(4):
    print(cube ** 3)
````

````python
# 103 — накопитель-список: пустой снаружи, append внутри
morse_dict = {"A": ".-", "L": ".-..", "F": "..-."}

morse_text = []
for letter in "ALFA":
    morse_text.append(morse_dict[letter])
    print(morse_text)

print(" ".join(morse_text))
````

Заметь, где стоит `morse_text = []` — **до** цикла. Если бы он стоял внутри, список обнулялся бы на каждом шаге.

````python
# 104 — накопитель-строка
VOWELS = "аеёиоуыэюя"

kinds = ""
for char in "привет":
    if char in VOWELS:
        kinds = kinds + "Г"
    else:
        kinds += "С"

print(kinds)
````

````python
# 105 — накопитель-число
total = 0
for i in range(1, 5):
    total = total + i
    print(i, total)
````

## Раздел 11. Функция

`def` ничего не выполняет. Он **запоминает** действия под именем, чтобы выполнить их потом,
столько раз, сколько нужно.

````python
# 110 — создаём функцию
def sm():
    return 2

print("функция создана, но ещё не выполнялась")
````

````python
# 111 — пока не вызовешь — ничего не произойдёт
print(sm())
print(sm())
````

````python
# 112 — параметр: значение приезжает снаружи
def give_me_squared_number(my_number):
    squared_number = my_number ** 2
    return squared_number

print(give_me_squared_number(9))
print(give_me_squared_number(3))
````

````python
# 113 — два параметра — порядок важен
def give_me_any_power(my_number, to_the_power):
    return my_number ** to_the_power

print(give_me_any_power(2, 10))
print(give_me_any_power(10, 2))
````

`my_number` и `to_the_power` — **параметры**, имена внутри функции.
`2` и `10` в вызове — **аргументы**, конкретные значения. Связываются они по порядку.

````python
# 114 — снаружи имя одно, внутри другое — значение то же
def hello(a):
    return a * 10000

number = 2
print(hello(number))
````

````python
# 115 — print ПОКАЗЫВАЕТ, return ОТДАЁТ
def hello_print():
    print("Hi")

def hello_return():
    return "Hi"

a = hello_print()
b = hello_return()

print("a =", a)
print("b =", b)
````

Обе функции как будто «дали Hi». Но `hello_print` только напечатала его на экран,
а отдала наружу **`None`**. Взять её результат и продолжить счёт нельзя.

````python
# 116 — функция без return возвращает None
def hello():
    print("Hi")

result = hello()
print(result)
print(type(result))
````

````python
# 117 — а с return результат можно считать дальше
def hello(a):
    print("Hi")
    return a * 10000

result = hello(2)
print(result + 1)
````

````python
# 118 — результат одной функции можно сразу отдать другой
def numbers_add(x, y, z):
    return x + y + z

print(numbers_add(10, 20, 30))
print(numbers_add(numbers_add(1, 1, 1), 20, 30))
````

````python
# 119 — наша функция из градиента: скользим от a к b
def mix(a, b, k):
    return a + (b - a) * k

print(mix(0, 100, 0))
print(mix(0, 100, 0.5))
print(mix(0, 100, 1))
````

````python
# 120 — функция, которая зовёт другую функцию
def between(color1, color2, k):
    r = mix(color1[0], color2[0], k)
    g = mix(color1[1], color2[1], k)
    b = mix(color1[2], color2[2], k)
    return (r, g, b)

color_a = (0.98, 0.85, 0.2)
color_b = (0.16, 0.2, 0.36)

print(between(color_a, color_b, 0))
print(between(color_a, color_b, 0.5))
print(between(color_a, color_b, 1))
````

В последней строчке вместо ровного `0.16` вылезло `0.16000000000000003`.
Это не ошибка функции: компьютер хранит дробные числа приблизительно.
Для цвета такая разница незаметна, а когда нужно красиво — есть `round(x, 4)`, как в `week1/circle.py`.

````python
# 121 — функция может изменить словарь, который ей дали
def check_white(a):
    if "white" not in a:
        a["white"] = -1.0
        print("ШАЛОСТЬ УДАЛАСЬ!")
    else:
        print("УЖЕ БЫЛО!")

a = {"red": 0, "green": 1}
check_white(a)
print(a)
check_white(a)
print(a)
````

````python
# 122 — а число так поменять не выйдет
def add_one(n):
    n = n + 1
    return n

x = 5
add_one(x)
print(x)
print(add_one(x))
````

Разница важная. Словарь и список функция меняет **на месте** — снаружи это видно.
Число и строку поменять нельзя, можно только посчитать новое значение и **вернуть** его.
Поэтому `velocity[0] = -velocity[0]` в мяче работает, а `n = n + 1` наружу не выходит.

````python
# 123 — return немедленно выходит из функции
def text_to_morse(text):
    morse_dict = {"A": ".-", "L": ".-..", "F": "..-."}
    result = []
    for letter in text:
        if letter not in morse_dict:
            print("ВВЕДИ НОРМАЛЬНЫЙ ТЕКСТ НА АНГЛИЙСКОМ!")
            return
        result.append(morse_dict[letter])
    return " ".join(result)

print(text_to_morse("ALFA"))
print(text_to_morse("АЛЬФА"))
````

Во втором вызове до `" ".join(...)` дело не дошло: голый `return` вышел сразу и отдал `None`.
Поэтому на экране и напечаталось `None`.

**Где мы это видели:** `week1/our_functions.py` и `week1/04_functions.py`; `week1/05_gradient.py` — `mix` и `between`;
`week1/morse_aux.py` — ранний `return`; `review.ipynb` — `check_white`.

## Раздел 12. Разбираем наш собственный код

````python
# 124 — velocity[0] = -velocity[0] — из прыгающего мяча
velocity = [4, 3]
print(velocity[0])
print(-velocity[0])

velocity[0] = -velocity[0]
print(velocity)
````

````python
# 125 — transitions[pair] += 1 — из марковской цепи
kinds = "СГСГГС"
transitions = {"ГГ": 0, "ГС": 0, "СГ": 0, "СС": 0}

for i in range(len(kinds) - 1):
    pair = kinds[i] + kinds[i + 1]
    print(i, pair)
    transitions[pair] += 1

print(transitions)
````

````python
# 126 — film["title_clean"].lower() — из скачивания постеров
film = {"title_clean": "Citizen Kane", "year": 1941, "rank": 1}

print(film["title_clean"])
print(film["title_clean"].lower())
print(str(film["rank"]).zfill(4))
````

````python
# 127 — и собираем имя файла целиком
title = ""
for char in film["title_clean"].lower():
    if char.isalnum():
        title += char
    elif char == " ":
        title += "_"

print(title)
print(str(film["rank"]).zfill(4) + "_" + title + "_" + str(film["year"]) + ".jpg")
````

````python
# 128 — choice(colors), когда в списке лежит список — из morellet.py
from random import choice

colors = [["red", "orange", "yellow"], "green", "blue", "gold", "black"]

for i in range(8):
    random_color = choice(colors)
    print(random_color, type(random_color))
````

Иногда `random_color` — строка, а иногда список. Отсюда и разное поведение рисунка.
