```python
def hello():
    print("Hi")
    return "Hi"

result = hello()

print(result)        # None
print(type(result))
```

```python
def hello(a):
    print("Hi")
    return a * 10000

result = hello(2)

print(result)        # None
print(type(result))
```

```python
def hello(a, b, c):
    print("Hi")
    return a * b * c * 10000

result = hello(2, 5, 6)

print(result)        # None
print(type(result))
```
