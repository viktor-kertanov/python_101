# Множества в Пайтон

```python
a = [1,2,3,4,5] #список
b = {"red": "красный", "blue": "синий"} #словарь
c = (128, 128, 128) #кортеж (tuple)
d = {1, 2, 5} # множество (set)
```

```python
def look(var):
    print(f"{type(var)} // {var}")
```

```python
look(a)
look(b)
look(c)
look(d)
```

```python
e = {11, 22, 33, 44, 44, 44, 55, 55, 66}
f = {1, 2, 3, 4, 4, 4, 5, 5, 11, 22, 33, 44, 55}
look(e)
look(f)
```

```python
common = e & f
look(common)
```

```python
all_elements = e | f
look(all_elements)
```

```python
only_e = e - f
look(only_e)
```

```python
only_f = f - e
look(only_f)
```

```python
not_in_both  = e ^ f
look(not_in_both)
```

```python
look(a)
```

```python
a_converted = set(a)
look(a_converted)
```

```python
colors = {(128, 128, 128), (0, 0, 0), (255, 255, 255), (128, 128, 128)}
look(colors)
```
