# Python модуль datetime

```python
from datetime import datetime, date, timedelta, timezone
```

```python
# Текущее время

now = datetime.now()
print(now)
```

```python
type(now)
```

```python
print(now.year)
print(now.month)
print(now.day)
print(now.hour)
print(now.minute)
print(now.second)
```

```python
#Только дату
today = date.today()
print(today)
```

```python
# Создать конкретную дату
lesson = datetime(2026, 9, 24, 19, 00)
print(lesson)
```

```python
# Задать только день
day = datetime(2026, 9, 24)
print(day)
```

# Сравнение двух дат

```python
a = datetime(2026, 1, 1) #дата номер один
b = datetime(2026, 9, 24) #дата номер два
```

```python
print(a > b)
print(b > a)
print(b != a)
```

```python
delta = b - a
```

```python
type(delta)
```

```python
print(delta.days)
```

```python
three_days = timedelta(days=3)
future = datetime.now() + three_days
print(future)
```

```python
text = future.strftime("%d / %m / %y :: %H:%M:%S")
print(text)
```

```python
my_text_date = "24=09=26"
d = datetime.strptime(my_text_date, "%d=%m=%y")
print(d)
```

```python
type(d)
```

# Кол-во секунд

```python
start_date = datetime(1989, 7, 19, 1, 20)
now = datetime.now()

seconds_i_live = (now - start_date).total_seconds()
print(seconds_i_live)
print(f"{seconds_i_live:,}")
print(f"{seconds_i_live:,.0f}")
print(f"{seconds_i_live:,.0f}".replace(",", " "))
```

```python
seconds_i_live / 3600 / 24 / 365
```
